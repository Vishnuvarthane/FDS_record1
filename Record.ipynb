Experiment 6: Time Series Analysis of Website Traffic

```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
from statsmodels.tsa.seasonal import seasonal_decompose

# Create data directly in the code
data = pd.DataFrame({
    'Date': pd.date_range(start='2025-01-01', periods=28, freq='D'),
    'Page_Views': [
        100, 120, 115, 130, 125, 150, 180,
        110, 125, 120, 135, 130, 155, 190,
        115, 130, 125, 140, 135, 160, 200,
        120, 135, 130, 145, 140, 170, 210
    ]
})

# Set Date as index
data.set_index('Date', inplace=True)

# 1. Time Series Decomposition
decomposition = seasonal_decompose(
    data['Page_Views'], model='additive', period=7
)

fig = decomposition.plot()
fig.set_size_inches(10, 8)
plt.suptitle('Time Series Decomposition', y=1.02)
plt.show()

# 2. Moving Average
data['Page_Views_MA'] = data['Page_Views'].rolling(window=7).mean()

plt.figure(figsize=(10, 4))
plt.plot(data.index, data['Page_Views'], label='Page Views')
plt.plot(
    data.index,
    data['Page_Views_MA'],
    label='7-Day Moving Average',
    color='orange'
)
plt.title('Page Views with Moving Average')
plt.xlabel('Date')
plt.ylabel('Page Views')
plt.legend()
plt.grid(True)
plt.show()

# 3. Anomaly Detection using Z-score
data['Page_Views_Z'] = (
    data['Page_Views'] - data['Page_Views'].mean()
) / data['Page_Views'].std()

anomalies = data[data['Page_Views_Z'].abs() > 3]

plt.figure(figsize=(10, 4))
plt.plot(
    data.index,
    data['Page_Views'],
    label='Page Views',
    color='C0',
    alpha=0.8
)
plt.scatter(
    anomalies.index,
    anomalies['Page_Views'],
    color='red',
    label='Anomalies',
    zorder=5
)
plt.title('Page Views Anomaly Detection')
plt.xlabel('Date')
plt.ylabel('Page Views')
plt.legend()
plt.grid(True)
plt.show()

# Display the data
print(data)
```

Experiment 7:. Random Sampling and Sampling Distribution

```python
import matplotlib.pyplot as plt
import numpy as np
population_mean = 50
population_std = 10
population_size = 100000
population = np.random.normal(
    population_mean, population_std, population_size
)
sample_sizes = [30, 50, 100]
num_samples = 1000
sample_means = {}
for size in sample_sizes:
    sample_means[size] = []
    for _ in range(num_samples):
        sample = np.random.choice(population, size=size, replace=False)
        sample_means[size].append(np.mean(sample))
plt.figure(figsize=(12, 8))
for i, size in enumerate(sample_sizes):
    plt.subplot(len(sample_sizes), 1, i + 1)
    plt.hist(sample_means[size], bins=30, alpha=0.7, label=f"Sample Size {size}")
    plt.axvline(
        np.mean(population),
        color="red",
        linestyle="dashed",
        linewidth=1.5,
        label="Population Mean",
    )
    plt.title(f"Sampling Distribution (Sample Size {size})")
    plt.xlabel("Sample Mean")
    plt.ylabel("Frequency")
    plt.legend()
plt.tight_layout()
plt.show()

```

Lab Experiment 8: Conduct Z test for the Data Given using Python Objective

```python
import scipy.stats as stats
sample_data = np.array([
    152, 148, 151, 149, 147, 153, 150, 148, 152, 149,
    151, 150, 149, 152, 151, 148, 150, 152, 149, 150,
    148, 153, 151, 150, 149, 152, 148, 151, 150, 153
])
population_mean = 150
sample_mean = np.mean(sample_data)
sample_std = np.std(sample_data, ddof=1)
n = len(sample_data)
z_statistic = (sample_mean - population_mean) / (sample_std / np.sqrt(n))
p_value = 2 * (1 - stats.norm.cdf(np.abs(z_statistic)))
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Z-Statistic: {z_statistic:.4f}")
print(f"P-Value: {p_value:.4f}")
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The average weight is significantly different from 150 grams.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in average weight from 150 grams.")
```

Exp No:12 Experiment to detect outliers in a given data set. Description: Understand the procedure to identify the outliers in a given dataset

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
%matplotlib inline
array = np.random.randint(1, 100, 16)
print("Original Array:", array)
print("Mean:", array.mean())
print("25th Percentile:", np.percentile(array, 25))
print("50th Percentile:", np.percentile(array, 50))
print("75th Percentile:", np.percentile(array, 75))
print("100th Percentile:", np.percentile(array, 100))
def outDetection(array):
    sorted(array)
    Q1, Q3 = np.percentile(array, [25, 75])
    IQR = Q3 - Q1
    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)
    return lr, ur
lr, ur = outDetection(array)
print("Lower Range:", lr, "Upper Range:", ur)
sns.displot(array)
sns.distplot(array)
plt.show()
new_array = array[(array > lr) & (array < ur)]
print("Array after 1st filtering:", new_array)
sns.displot(new_array)
plt.show()
lr1, ur1 = outDetection(new_array)
print("New Lower Range:", lr1, "New Upper Range:", ur1)
final_array = new_array[(new_array > lr1) & (new_array < ur1)]
print("Final Array:", final_array)
sns.distplot(final_array)
plt.show()
```

```python
import numpy as np
import pandas as pd
import os

if not os.path.exists("Hotel_Dataset.csv"):
    mock_data = {
        'CustomerID': [1001, 1002, -5, 1004, 1005, 1006, 1001, 1008, 1009, 1010],
        'Age_Group': ['Young', 'Middle-Aged', 'Senior', 'Young', 'Middle-Aged', 'Senior', 'Young', 'Young', 'Senior', 'Middle-Aged'],
        'Age_Group.1': ['Young', 'Middle-Aged', 'Senior', 'Young', 'Middle-Aged', 'Senior', 'Young', 'Young', 'Senior', 'Middle-Aged'],
        'Hotel': ['Ibis', 'Ibys', 'Novotel', 'Ibis', 'Novotel', 'Ibis', 'Ibis', 'Ibis', 'Novotel', 'Ibis'],
        'FoodPreference': ['Veg', 'Vegetarian', 'veg', 'non-Veg', 'Veg', 'Non-Veg', 'Veg', 'Veg', 'Non-Veg', 'Veg'],
        'NoOfPax': [2, 3, 25, 1, 4, 2, 2, np.nan, 2, 0],
        'Bill': [1500, 2400, -100, 1200, 3100, np.nan, 1500, 1800, 2900, 2100],
        'EstimatedSalary': [50000, 60000, 45000, -5000, 80000, 95000, 50000, np.nan, 75000, 62000],
        'Rating(1-5)': [4, 5, 3, 4, np.nan, 2, 4, 5, 4, 3]
    }

    pd.DataFrame(mock_data).to_csv("Hotel_Dataset.csv", index=False)
    print("Generated mock Hotel_Dataset.csv successfully.")

# Load data
df = pd.read_csv("Hotel_Dataset.csv")

print("Original Data:")
print(df)

# 1. Check duplicate rows
print("\nDuplicated rows mask:")
print(df.duplicated())

df.info()

# 2. Remove duplicate rows
df.drop_duplicates(inplace=True)

print("\nAfter dropping duplicates:")
print(df)

# 3. Reset index
df.reset_index(drop=True, inplace=True)

print("\nIndex values:")
print(df.index.to_numpy())
print(df)

# 4. Remove redundant column
df.drop(columns=['Age_Group.1'], inplace=True)

print("\nAfter dropping redundant Age_Group.1 column:")
print(df)

# 5. Replace negative values with NaN
df.loc[df['CustomerID'] < 0, 'CustomerID'] = np.nan
df.loc[df['Bill'] < 0, 'Bill'] = np.nan
df.loc[df['EstimatedSalary'] < 0, 'EstimatedSalary'] = np.nan

print("\nAfter replacing negative values with NaN:")
print(df)

# 6. Replace invalid NoOfPax values with NaN
df.loc[(df['NoOfPax'] < 1) | (df['NoOfPax'] > 20), 'NoOfPax'] = np.nan

print("\nAfter replacing invalid NoOfPax with NaN:")
print(df)

# 7. Display unique values
print("\nUnique values:")
print("Age Group:", df['Age_Group'].unique())
print("Hotel:", df['Hotel'].unique())
print("Food Preference:", df['FoodPreference'].unique())

# 8. Correct inconsistent categorical values
df['Hotel'] = df['Hotel'].replace('Ibys', 'Ibis')

df['FoodPreference'] = df['FoodPreference'].replace({
    'Vegetarian': 'Veg',
    'veg': 'Veg',
    'non-Veg': 'Non-Veg'
})

# 9. Fill missing values
df['EstimatedSalary'] = df['EstimatedSalary'].fillna(
    round(df['EstimatedSalary'].mean())
)

df['NoOfPax'] = df['NoOfPax'].fillna(
    round(df['NoOfPax'].median())
)

df['Rating(1-5)'] = df['Rating(1-5)'].fillna(
    round(df['Rating(1-5)'].median())
)

df['Bill'] = df['Bill'].fillna(
    round(df['Bill'].mean())
)

# 10. Display final processed data
print("\nFinal Processed Data:")
print(df)
```

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, classification_report

# Load Iris dataset directly
iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['variety'] = iris.target_names[iris.target]

# Display dataset information
df.info()
print(df['variety'].value_counts())
print(df.head())

# Separate features and label
features = df.iloc[:, :-1].values
label = df.iloc[:, 4].values

# Split dataset into training and testing data
xtrain, xtest, ytrain, ytest = train_test_split(
    features, label, test_size=0.2, random_state=0
)

# Create and train KNN model
model_KNN = KNeighborsClassifier(n_neighbors=5)
model_KNN.fit(xtrain, ytrain)

# Calculate accuracy
print("Training Accuracy:", model_KNN.score(xtrain, ytrain))
print("Testing Accuracy:", model_KNN.score(xtest, ytest))

# Confusion matrix using test data
print("Confusion Matrix:")
print(confusion_matrix(ytest, model_KNN.predict(xtest)))

# Classification report using test data
print("Classification Report:")
print(classification_report(ytest, model_KNN.predict(xtest)))
```

Exp No:19 Experiment to understand Logistic Regression for a given data set.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

np.random.seed(42)

n_samples = 400

Age = np.random.randint(18, 60, size=n_samples)
EstimatedSalary = np.random.randint(15000, 150000, size=n_samples)

# Create Purchased values related to age and salary
probability = 1 / (
    1 + np.exp(-(-6 + 0.12 * (Age - 35) +
                 0.000025 * (EstimatedSalary - 75000)))
)

Purchased = np.random.binomial(1, probability)

df = pd.DataFrame({
    'Age': Age,
    'EstimatedSalary': EstimatedSalary,
    'Purchased': Purchased
})

features = df[['Age', 'EstimatedSalary']]
label = df['Purchased']

x_train, x_test, y_train, y_test = train_test_split(
    features, label, test_size=0.2,
    random_state=26, stratify=label
)

finalModel = make_pipeline(
    StandardScaler(),
    LogisticRegression()
)

finalModel.fit(x_train, y_train)

print("Training Accuracy:", finalModel.score(x_train, y_train))
print("Testing Accuracy:", finalModel.score(x_test, y_test))

y_pred = finalModel.predict(x_test)

print("Classification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

Exp No:20 Experiment to understand K-means clustering algorithm for a given data set.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
import os

# Generate sample data if the CSV file is missing
if not os.path.exists('Mall_Customers.csv'):
    np.random.seed(42)
    n_customers = 200

    mock_data = {
        'CustomerID': np.arange(1, n_customers + 1),
        'Genre': np.random.choice(['Male', 'Female'], n_customers),
        'Age': np.random.randint(18, 70, n_customers),
        'Annual Income (k$)': np.random.randint(15, 140, n_customers),
        'Spending Score (1-100)': np.random.randint(1, 100, n_customers)
    }

    df_mock = pd.DataFrame(mock_data)
    df_mock.to_csv('Mall_Customers.csv', index=False)
    print("Mall_Customers.csv generated successfully!")

# Read dataset
df = pd.read_csv('Mall_Customers.csv')

print(df.info())
print(df.head())

# Display pairplot
sns.pairplot(df)
plt.show()

# Select Annual Income and Spending Score
features = df.iloc[:, [3, 4]].values

# Apply K-Means clustering
model = KMeans(n_clusters=5, random_state=42, n_init=10)
model.fit(features)

# Add cluster labels
Final = df.iloc[:, [3, 4]].copy()
Final['label'] = model.labels_

print(Final.head())

# Plot clusters
sns.set_style("whitegrid")

sns.FacetGrid(Final, hue="label", height=6) \
    .map(plt.scatter, "Annual Income (k$)",
         "Spending Score (1-100)") \
    .add_legend()

plt.show()

# Elbow Method
features_el = df.iloc[:, [2, 3, 4]].values
wcss = []

for i in range(1, 10):
    model = KMeans(n_clusters=i, random_state=42, n_init=10)
    model.fit(features_el)
    wcss.append(model.inertia_)

# Plot Elbow graph
plt.plot(range(1, 10), wcss, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```
